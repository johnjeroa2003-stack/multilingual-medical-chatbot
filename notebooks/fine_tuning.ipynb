```python
from datasets import load_dataset
from transformers import AutoTokenizer
```

```python
dataset = load_dataset("lavita/MedQuAD")

print(dataset)
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    "google/flan-t5-small"
)
```

```python
sample_text = "What are symptoms of fever?"

tokens = tokenizer(
    sample_text,
    padding="max_length",
    truncation=True
)

print(tokens)
```

# Fine-Tuning Overview

This notebook demonstrates the preprocessing and tokenization workflow used for medical chatbot fine-tuning.

The FLAN-T5 model can be fine-tuned using medical question-answer datasets such as MedQuAD to improve healthcare-related response generation.
